#### - Q2 : **Ouverture du fichier txt**

```python
with open('../data/discours_candidat.txt') as f:
    discours = f.read()
```

#### - **Q3 : Traitements élémentaires** :  Tockenisation (spaCy), passage en minuscules, suppression de la ponctuation et des mots vides

```python
### Tokenisation, minuscule, gestion de la ponctuation et des mots vides
doc = nlp_fr(discours.lower())

tokens = [token for token in doc if not token.is_stop and not token.is_punct and not token.is_space]
print(tokens)
```

#### - **Q4 : Lemmatisation** (Spacy)

```python
### Lemmatisation
lemmas = [token.lemma_ for token in tokens]
print(lemmas)
```
